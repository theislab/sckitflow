```python
import sc_flow
```

```python
n_obs_pert = 10000
n_obs_ctrl = 5000
adata = sc_flow.data.sim.get_dummy_adata(n_obs_pert=n_obs_pert, n_obs_ctrl=n_obs_ctrl)
adata
```

```python
sc_flow.SCFlow.register_adata(
    adata,
    conditions={"drug": ("drugA", "drugB")},
    conditions_reps={"drug": "drug"},
    conditions_covariates=["paired_condition"],
    control_values_dict={"drug": "control"},
    view_on_condition_space=True,
    condition_state_key="paired_condition",
)
model_cond = sc_flow.SCFlow(method_id="cfm")
```

```python
sc_flow.SCFlow.register_adata(
    adata,
    conditions={"drug": ("drugA", "drugB")},
    conditions_reps={"drug": "drug"},
    conditions_covariates=["paired_condition"],
    control_values_dict={"drug": "control"},
)
model_cells = sc_flow.SCFlow(method_id="cfm")
```

```python
cond_tree = model_cond.dm.compile_adata(adata, sort=True)
cells_tree = model_cells.dm.compile_adata(adata, sort=True)
```
